```python
# --- librerias --- #

import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
import warnings

from pathlib import Path

# --- set cwd --- #
ROOT = Path().cwd().parent
os.chdir(ROOT)
pd.set_option("display.max_columns",None)
warnings.filterwarnings("ignore")
```

```python
# --- Leemos el archivo de labels --- #

labels = (
    pd.read_parquet(
        r'data\raw\labels\labels_27-09-2026.parquet', 
        engine='pyarrow',
    )
    .drop(
        columns=['strEstadoInicial'], 
    )
)

# --- Leemos la vista 360 --- #
v360_ini = (
    pd.read_parquet(
        r'data\raw\features\v_360\v_360_27-09-2026.parquet', 
        engine='pyarrow',   
    )
)
```

```python
# --- Realizamos un merge entre la vista 360 y los labels --- #

v360= (
    pd.merge(
        labels, 
        v360_ini, 
        how='left',
        on=['strPeriodo','strIdentificacion'],  
    )
    .drop(
        columns=['strIdentificacion', 'strPeriodo']
    )
)

v360.head()
```

```python
id_columns = ["skcliente",
              "Estado",
              "IndInactivo",
              "Regional",
              "Zona",
              "Oficina",
              "Area Conocimiento",
              "Habeas Data",
              "ColaboradorGECC",
              "Tipo Cliente",
              "Tipologia_Ciudad_Vinculacion",
              "Tipologia_Ciudad_Residencia",
              "Nombre Asociado",
              "CodDaneCiudadResidencia",
              "CiudadResidencia",
              "CodDaneCiudadHomologada",
              "CiudadHomologada",
              "dtmFechaInsercion",
              "dtmFechaActualizacionBI",
              "Alerta_ActualizacionDatos",
              "Fecha_Actualizacion_NPS_SAO",
              "Fecha_Actualizacion_NPS_MI",
              "TipoIdentificacion",
              "Turnos_En_Oficinas_UltimaOficinaVisitada_Ult12Meses",
              "Identificacion",
              "Periodo"]

PREFIJO = 'v360'

id_columns_pre = [f'{PREFIJO}_{col}' for col in id_columns] + ['strIdentificacion' , 'strPeriodo']

v360= (
    v360
    .drop(
        columns=id_columns_pre
    )
)

print(v360.shape[1])
```

## **Columnas CeroPapel -- Producto_Recomendado**

```python
cero_papel_columns = [c for c in v360.columns if c.strip().__contains__("CeroPapel")]
producto_recomendado_columns = [c for c in v360.columns if c.strip().__contains__("Producto")]

columnas_eliminar = cero_papel_columns + producto_recomendado_columns

v360= (
    v360
    .drop(
        columns=columnas_eliminar
    )
)
```

```python
v360.info()
```

## **Columnas Con un alto (%) de nulos**

```python
# Se reemplazan vacios por np.nan para que pandas los reconozca como nulos y no como un string

VALORES_NULOS = ["", " ", "  ", "nan", "NaN", "NULL", "null", 
                 "N/A", "n/a", "NA", "ninguno", "NINGUNO", "-","Sin información"]

v360 = v360.replace(VALORES_NULOS,np.nan)

# Proporción de nulos x columna

pp_nulls = v360.isnull().mean()*100

threshold = 30 ## Columnas con el 30% o mas de nulos -- Eliminar

eliminar_80_nulos = pp_nulls.loc[pp_nulls >= threshold].index

v360 = v360[[c for c in v360.columns if c not in eliminar_80_nulos]]

v360.shape
```

## **Columnas Constantes**

```python
col_constantes = [c for c in v360.columns 
                  if v360[c].nunique(dropna=True) <= 1]

v360 = v360[[c for c in v360.columns if c not in col_constantes]]
```

## **Columnas Cuasi Constantes**

```python
cols_cuasi_constantes = [c for c in v360.columns
                         if v360[c].value_counts(normalize=True).iloc[0] >= 0.95]

col_cuasi_eliminar = [c for c in  cols_cuasi_constantes if c != 'IndicadorReactivado']

# --- Se validan tasas de reactivacion por valor --- #
def evaluar_candidatas(df, candidatas, target='IndicadorReactivado',
                       min_pos=50, lift_alto=2.0):
    base = df[target].mean()
    filas = []

    for col in candidatas:
        g = df.groupby(col, dropna=False)[target].agg(['size', 'sum'])
        g.columns = ['n', 'pos']
        moda = g['n'].idxmax()
        minoria = g.drop(index=moda)

        n_min   = minoria['n'].sum()
        pos_min = minoria['pos'].sum()
        tasa    = pos_min / n_min if n_min else 0
        lift    = tasa / base if base else 0

        filas.append({
            'columna': col,
            'pct_moda': g['n'].max() / len(df),
            'n_minoria': n_min,
            'pos_minoria': pos_min,
            'tasa_minoria': tasa,
            'lift_minoria': lift,
            'rescatar': (pos_min >= min_pos) and
                        (lift >= lift_alto)
        })

    return pd.DataFrame(filas).sort_values('lift_minoria', ascending=False)

evaluacion_cuasi = evaluar_candidatas(df=v360, candidatas=col_cuasi_eliminar,)

evaluacion_cuasi.loc[evaluacion_cuasi['rescatar'] == True].shape[0]
```

```python
col_cuasi_eliminar = evaluacion_cuasi.loc[evaluacion_cuasi['rescatar'] == False, 'columna'].values

v360 = v360.drop(
    columns=col_cuasi_eliminar,
)
```

```python

## Solo para las columnas numericas -- Se imputa temporalmente con mediana para calcular corr

v_360_num = v360.select_dtypes("number")
v_360_num_imp = v_360_num.fillna(v_360_num.median())

## Matriz de correlación

corr_matrix = v_360_num_imp.corr(method='spearman').abs()

UMBRAL_CORR = 0.95

## Columnas con correlación alta

upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

cols_alta_corr = [col for col in upper.columns if any(upper[col] > UMBRAL_CORR)]

cols_alta_corr
```

```python
## Eliminar las columnas alta correlación

v360 =  v360.drop(columns=cols_alta_corr)

v360.head()
```

```python
# --- Columnas eliminar x logica --- #

col_eliminar_logica = ["v360_Tipo_Cliente_Asociado",
                       "v360_UltimoEventoRecreacionAsistio_Histo",
                       "v360_DebitoAutomatico",
                       "v360_TIENE EL DEBITO AUTOMATICO AUTORIZADO",
                       "v360_Corte_Facturacion_Asociado",
                       "v360_Profesion",
                       ]

v360 = (
    v360
    .drop(
        columns=col_eliminar_logica, 
    )
)
```

## **Documentacion**

```python
## Se genera un DataFrame -- documentación

v360_ini = v360_ini.replace(VALORES_NULOS,np.nan)

doc_vista_360 = pd.DataFrame({"Nombre_Columna":v360_ini.columns.tolist(),
                              "Tipo_Dato":v360_ini.dtypes,
                              "Cantidad_Nulos":v360_ini.isnull().mean().values,
                              }).reset_index(drop=True)

# --- Dataframe valores moda --- #

wrapper = {}

for c in v360_ini.columns:
    VALOR = round(v360_ini[c].value_counts(dropna=False,normalize=True).iloc[0], 2)
    wrapper[c] = VALOR
    
moda = (
    pd.DataFrame(
        {
            'Nombre_Columna': wrapper.keys(), 
            'FrecuenciaModa': wrapper.values()
        }
    )
)

doc_vista_360 = (
    doc_vista_360
    .merge(
        moda, 
        how='left', 
        on='Nombre_Columna'
    )
)

## Se generan comentarios

condiciones = [doc_vista_360["Nombre_Columna"].isin(id_columns_pre),
               doc_vista_360["Nombre_Columna"].isin(cero_papel_columns),
               doc_vista_360["Nombre_Columna"].isin(producto_recomendado_columns),
               doc_vista_360["Nombre_Columna"].isin(eliminar_80_nulos),
               doc_vista_360["Nombre_Columna"].isin(col_constantes),
               doc_vista_360["Nombre_Columna"].isin(col_cuasi_eliminar),
               doc_vista_360["Nombre_Columna"].isin(col_eliminar_logica),
               doc_vista_360["Nombre_Columna"].isin(cols_alta_corr)]

valores = ["Columna Identificadora",
           "Columna Cero Papel - No util",
           "Columna Producto Recomendado - No Util",
           "Alta Proporcion Nulos",
           "Columna Constante",
           "Columna Cuasi-Constante (90% mismo valor)",
           "Columna Eliminada por logica",
           "Columnas con 95% de correlacion"]

doc_vista_360["Decision"] = np.select(condlist=condiciones,
                                      choicelist=valores,
                                      default="Tener en cuenta")

doc_vista_360["Decision"].value_counts(normalize=True,
                                       dropna=False)
```

```python
columnas_str = doc_vista_360[(doc_vista_360["Decision"] == "Tener en cuenta") &
                             (doc_vista_360["Tipo_Dato"] == "str")]["Nombre_Columna"]

## Cambiar tipos Identificacion y Periodo

doc_vista_360.loc[doc_vista_360["Nombre_Columna"].isin(["Identificacion",
                                                        "Periodo"]),"Tipo_Dato"] = "int"

## Cambiar Tipo_Dato -- binaria

columnas_binarias = [col for col in v360.columns if v360[col].nunique() <= 2]

doc_vista_360.loc[doc_vista_360["Nombre_Columna"].isin(columnas_binarias),
                  "Tipo_Dato"] = "binaria"
```

```python
doc_vista_360["Tipo_Dato"] = doc_vista_360["Tipo_Dato"].astype(str)

# Reemplazar object with str

doc_vista_360["Tipo_Dato"] = doc_vista_360["Tipo_Dato"].replace("object",
                                   "str")
```
